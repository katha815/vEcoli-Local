```python
from pathlib import Path

input_file = Path('/user/home/il22158/work/vEcoli/reading/imported/Single_KO_RNA_names_formatted.txt')
output_dir = Path('/user/home/il22158/work/vEcoli/reading/imported/')
lines = input_file.read_text().splitlines()

chunk_size, remainder = divmod(len(lines), 5)
parts = []
start = 0
for index in range(5):
    stop = start + chunk_size + (1 if index < remainder else 0)
    parts.append((start + 1, stop, lines[start:stop]))
    start = stop

for index, (start_line, end_line, part) in enumerate(parts, start=1):
    output_file = output_dir / f'part{index}_{start_line}-{end_line}.txt'
    output_file.write_text('\n'.join(part) + '\n')
```
